```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
data1 = np.linspace(0, 10, 20)
```

```python
data1
```

```python
data2 = data1 ** 2
```

```python
plt.subplots()
```

```python
my_fig, my_axes = plt.subplots()
```

```python
type(my_fig)
```

```python
type(my_axes)
```

```python
my_fig, my_axes = plt.subplots()
my_axes.plot(data1, data2, color="#1E0703")
my_axes.plot(data2, data1, color="#53CA0A")
plt.show()
```

```python
(new_fig, new_axes) = plt.subplots()
new_axes.plot(data1, data1 + 2, color="blue", linewidth=1.0)
new_axes.plot(data1, data1 + 3, color="red", linewidth=4.0)
new_axes.plot(data1, data1 + 4, color="green", linestyle="-.")
new_axes.plot(data1, data1 + 5, color="black", linestyle="--", marker="o", markersize=8, markerfacecolor="red")
new_axes.plot(data1, data1 + 6, color="black", linestyle="--", marker="+", markersize=6)
plt.show()
```

```python
plt.scatter(data1, data2)
plt.show()
```

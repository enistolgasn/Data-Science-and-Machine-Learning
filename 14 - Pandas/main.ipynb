Multi Index

```python
first_index = ["Simpson", "Simpson", "Simpson", "South Park", "South Park", "South Park"]
```

```python
innder_index = ["Homer", "Bart", "Marge", "Cartman", "Kenny", "Kyle"]
```

```python
zipped_index = list(zip(first_index, innder_index))
```

```python
zipped_index
```

```python
zipped_index = pd.MultiIndex.from_tuples(zipped_index)
```

```python
zipped_index
```

```python
sample_values = np.ones((6, 2))
```

```python
multi_index_df = pd.DataFrame(sample_values, index=zipped_index, columns=["Salary", "Seniority"])
```

```python
multi_index_df
```

```python
multi_index_df.loc["Simpson"]
```

```python
multi_index_df.loc["Simpson"].loc["Homer"]
```

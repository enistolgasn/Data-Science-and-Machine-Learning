```python
import numpy as np
```

```python
import pandas as pd
```

```python
weather_df = pd.read_excel("weather.xlsx")
```

```python
weather_df
```

```python
# head method returns the first 5 rows of the dataframe by default. You can also specify the number of rows to return by passing an integer as an argument to the head method. For example, weather_df.head(10) will return the first 10 rows of the dataframe.
weather_df.head()
```

```python
# tail method returns the last 5 rows of the dataframe by default. You can also specify the number of rows to return by passing an integer as an argument to the tail method. For example, weather_df.tail(10) will return the last 10 rows of the dataframe.
weather_df.tail()
```

```python
# info method prints a summary of the dataframe, including the number of non-null values, the data types of each column, and the memory usage of the dataframe.
weather_df.info()
```

```python
# describe method generates descriptive statistics that summarize the central tendency, dispersion, and shape of a dataset’s distribution, excluding NaN values. It provides information such as count, mean, standard deviation, minimum, maximum, and quartile values for each numerical column in the dataframe.
weather_df.describe()
```

```python
# count method returns the number of non-null values in each column of the dataframe. It can be useful for identifying missing data in the dataframe.
weather_df.count()
```

```python
# isna method returns a dataframe of the same shape as the original dataframe, but with boolean values indicating whether each element is missing (NaN) or not. It can be useful for identifying missing data in the dataframe.
weather_df.isna()
```
